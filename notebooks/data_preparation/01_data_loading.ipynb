# Data Loading and Initial Exploration

This notebook handles the initial data loading and exploration steps for the fraud detection model.

## Setup Environment

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import os
```

## Load Dataset

```python
data = pd.read_parquet('../data/credit_card_transactions.parquet')
print(f"Loaded dataset with shape: {data.shape}")
```

## Initial Data Exploration

```python
# Display basic information about the dataset
print("Dataset Info:")
data.info()

print("\nSample of data:")
data.head()
```

## Data Distribution Analysis

```python
# Analyze class distribution
print("Class distribution:")
print(data['is_fraud'].value_counts(normalize=True))

# Temporal distribution
ym_stats = data[['trans_date_trans_time','state','is_fraud','trans_num']]
ym_stats['ym'] = data.trans_date_trans_time.dt.year.astype(str) + data.trans_date_trans_time.dt.month.astype(str).str.zfill(2)
monthly_stats = ym_stats.groupby(['ym']).agg({'is_fraud':sum, 'trans_num': 'nunique'})
print("\nMonthly statistics:")
print(monthly_stats)
```
